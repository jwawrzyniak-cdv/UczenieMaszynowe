### 2.4 Pierwszy MLP

```python
def plot_history(history, title=""):
    h = pd.DataFrame(history.history)
    fig, axes = plt.subplots(1, 2, figsize=(11, 3.6))
    h[["loss", "val_loss"]].plot(ax=axes[0])
    axes[0].set_title("Funkcja straty " + title)
    axes[0].set_xlabel("epoka")
    h[["accuracy", "val_accuracy"]].plot(ax=axes[1])
    axes[1].set_title("Dokładność " + title)
    axes[1].set_xlabel("epoka")
    plt.tight_layout()
    plt.show()
```
